### Ingest Data from s3

```python
table=[{
    "file_name":"Customers.csv",
    "table_name":"customers"
},{
    "file_name":"Products.csv",
    "table_name":"products"
},
{
    "file_name":"Sales.csv",
    "table_name":"sales"
},{
    "file_name":"Stores.csv",
    "table_name":"stores"
},{
    "file_name":"Data_Dictionary.csv",
    "table_name":"data_dictionary"
},{
    "file_name":"Exchange_Rate.csv",
    "table_name":"exchange_rate"
}]


for t in table:
    df = spark.read.csv(
    f"s3://electronics-retailer-bucket/bronze/{t['file_name']}",
    header=True,
    inferSchema=True
)
    df = df.toDF(*[c.replace(" ", "_") for c in df.columns])
    df.write.format("delta").mode("overwrite").saveAsTable(f"01_prod_bronze.raw.{t['table_name']}")
     
```
