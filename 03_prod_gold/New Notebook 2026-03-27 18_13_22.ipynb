```sql
CREATE OR REPLACE TABLE 03_prod_gold.analytics.sales_datacube AS

SELECT
    -- ======================
    -- FACT KEYS & ORDER INFO
    -- ======================
    COALESCE(f.order_number, -1)        AS order_number,
    COALESCE(f.line_item, -1)           AS line_item,

    COALESCE(f.order_date, DATE '1900-01-01')    AS order_date,
    COALESCE(f.delivery_date, DATE '1900-01-01') AS delivery_date,

    -- ======================
    -- DATE DIMENSION
    -- ======================
    COALESCE(d.year, -1)        AS year,
    COALESCE(d.month, -1)       AS month,
    COALESCE(d.day, -1)         AS day,

    -- ======================
    -- CUSTOMER DIMENSION
    -- ======================
    COALESCE(c.customerkey, -1)      AS customer_key,
    COALESCE(c.gender, 'unknown')    AS gender,
    COALESCE(c.city, 'unknown')      AS customer_city,
    COALESCE(c.state, 'unknown')     AS customer_state,
    COALESCE(c.country, 'unknown')   AS customer_country,
    COALESCE(c.continent, 'unknown') AS continent,
    COALESCE(c.birthday, DATE '1900-01-01') AS birthday,

    -- ======================
    -- STORE DIMENSION
    -- ======================
    COALESCE(s.storekey, -1)        AS store_key,
    COALESCE(s.country, 'unknown')  AS store_country,
    COALESCE(s.state, 'unknown')    AS store_state,
    COALESCE(s.square_meters, -1)   AS store_size,
    COALESCE(s.open_date, DATE '1900-01-01') AS store_open_date,

    -- ======================
    -- PRODUCT DIMENSION
    -- ======================
    COALESCE(p.productkey, -1)        AS product_key,
    COALESCE(p.product_name, 'unknown') AS product_name,
    COALESCE(p.brand, 'unknown')      AS brand,
    COALESCE(p.color, 'unknown')      AS color,
    COALESCE(p.subcategory, 'unknown') AS subcategory,
    COALESCE(p.category, 'unknown')    AS category,

    COALESCE(p.unit_cost_usd, 0.0)    AS unit_cost_usd_dim,
    COALESCE(p.unit_price_usd, 0.0)   AS unit_price_usd_dim,

    -- ======================
    -- FACT MEASURES
    -- ======================
    COALESCE(f.quantity, 0)          AS quantity,
    COALESCE(f.currency_code, 'unknown') AS currency_code,

    COALESCE(f.unit_price_usd, 0.0)  AS unit_price_usd,
    COALESCE(f.revenue_usd, 0.0)     AS revenue_usd,
    COALESCE(f.cost_usd, 0.0)        AS cost_usd,
    COALESCE(f.profit_usd, 0.0)      AS profit_usd,

    COALESCE(f.delivery_days, 0)     AS delivery_days,

    -- ======================
    -- DERIVED FIELDS
    -- ======================
    CASE 
        WHEN f.storekey IS NULL THEN 'online'
        ELSE 'store'
    END AS sales_channel
 
FROM `03_prod_gold`.analytics.fact_sales f

-- DATE JOIN
LEFT JOIN `03_prod_gold`.analytics.dim_date d
    ON f.order_date = d.date

-- CUSTOMER JOIN
LEFT JOIN `03_prod_gold`.analytics.dim_customers c
    ON f.customerkey = c.customerkey

-- STORE JOIN
LEFT JOIN `03_prod_gold`.analytics.dim_stores s
    ON f.storekey = s.storekey

-- PRODUCT JOIN
LEFT JOIN `03_prod_gold`.analytics.dim_products p
    ON f.productkey = p.productkey;
```

```sql
select * from 03_prod_gold.analytics.sales_datacube
```
